```python
import xarray as xr

ds = xr.open_dataset("wrf_daily_EPA_d01_20260529_20260611.nc", engine="netcdf4")
print(ds)
print("\nCoordinates sample:")
print("lat range:", float(ds.lat.min()), "to", float(ds.lat.max()))
print("lon range:", float(ds.lon.min()), "to", float(ds.lon.max()))
print("time:", ds.time.values)
print("\nVariable stats:")
for var in ds.data_vars:
    print(f"{var}: min={float(ds[var].min()):.2f}, max={float(ds[var].max()):.2f}, units={ds[var].attrs.get('units', 'N/A')}")
```
